#### ✈️ Flight Booking & Revenue Analytics System
📄 Problem Statement

Design and implement a Flight Booking Analytics System that processes booking data and generates meaningful business, operational, and revenue insights, similar to real-world airline analytics platforms.

The system should handle booking-level data and compute key performance indicators (KPIs) that help stakeholders understand revenue performance, customer behavior, and operational efficiency.

🎯 Objectives  

1️⃣ Revenue Analytics  
Calculate total revenue per flight (only confirmed bookings)  
Calculate revenue per travel class (Economy, Business, First)  
Identify flights generating revenue greater than a threshold (e.g., 50,000)  
Calculate revenue distribution by class within each flight  

2️⃣ Booking Insights  
Compute passenger count per flight (confirmed bookings only)  
Identify the most booked flight  
Calculate cancellation rate per flight  
(Cancelled bookings / Total bookings)  

3️⃣ Business Intelligence  
Identify the most profitable travel class  
Calculate average ticket price per flight (confirmed bookings)  

4️⃣ Operational Metrics  
Calculate cancellation count per flight  
Calculate total bookings per flight  

📦 Expected Output Structure  
{
    "revenue_per_flight": {...},  
    "revenue_per_class": {...},  
    "revenue_by_class_per_flight": {...},   
    "high_revenue_flights": [],  
    "passenger_count_per_flight": {...},  
    "most_booked_flight": (...),  
    "cancellation_rate_per_flight": {...},  
    "most_profitable_class": (...),  
    "avg_ticket_price_per_flight": {...},  
    "cancellation_count_per_flight": {...},   
    "total_bookings_per_flight": {...}    
}

```python
bookings = [
    ("F1", "P1", "Economy", 5000, "Confirmed"),
    ("F2", "P2", "Business", 20000, "Confirmed"),
    ("F1", "P3", "Economy", 5000, "Cancelled"),
    ("F3", "P4", "First", 40000, "Confirmed"),
    ("F2", "P5", "Economy", 7000, "Confirmed"),
    ("F1", "P6", "Business", 15000, "Confirmed"),
    ("F3", "P7", "First", 45000, "Confirmed"),
    ("F3", "P12", "First", 40000, "Cancelled"),
    ("F3", "P8", "First", 4000, "Confirmed"),
    ("F1", "P11", "Economy", 15000, "Cancelled"),
]

total_bookings_per_flight = {}
revenue_by_flight = {}
revenue_by_flight_class = {}
revenue_by_class = {}
flight_generating_max_revenue = []
passenger_count_per_flight = {}
avg_ticket_price_per_flight = {}
flight_avg_price = {}
flights_with_highest_cancellation_rate = {}
cancellation_cnt_per_flight = {}
cancellation_rate = {}



for flight,passenger,class_,amt,status in bookings:
    
    if flight not in total_bookings_per_flight:
        total_bookings_per_flight[flight] = 0
        
    total_bookings_per_flight[flight] += 1
    
    if status.lower() == 'confirmed':
        
        if flight not in revenue_by_flight:
            revenue_by_flight[flight]  = 0
        
        revenue_by_flight[flight] += amt
        
        if flight not in revenue_by_flight_class:
            revenue_by_flight_class[flight] = {}

        if class_ not in revenue_by_flight_class[flight]:
            revenue_by_flight_class[flight][class_] = 0
            
        revenue_by_flight_class[flight][class_] += amt
        
        if class_ not in revenue_by_class:
            revenue_by_class[class_] = 0
            
        revenue_by_class[class_] += amt
        
        if flight not in passenger_count_per_flight:
            passenger_count_per_flight[flight] = 0
            
        passenger_count_per_flight[flight] += 1
        
        if flight not in avg_ticket_price_per_flight:
            avg_ticket_price_per_flight[flight] = []
        
        avg_ticket_price_per_flight[flight].append(amt)
        
    if status.lower() == 'cancelled':
        
        if flight not in cancellation_cnt_per_flight:
            cancellation_cnt_per_flight[flight] = 0
            
        cancellation_cnt_per_flight[flight] += 1
        
        
for k,v in revenue_by_flight.items():
     if v > 50000:
            flight_generating_max_revenue.append((k,v))
        
for k,v in avg_ticket_price_per_flight.items():
    avg_price = sum(v)/len(v)
    flight_avg_price[k] = avg_price
        
        
for flight in total_bookings_per_flight:
    cancelled = cancellation_cnt_per_flight.get(flight,0)
    total = total_bookings_per_flight[flight]
    cancellation_rate[flight] = cancelled/total
        
most_booked_flight = max(passenger_count_per_flight.items(), key=lambda x:x[1])
most_profitable_class = max(revenue_by_class.items(), key=lambda x:x[1])

output_dict = {
    # revenue analytics
    "revenue per flight" : revenue_by_flight,
    "revenue per class" : revenue_by_class,
    "revenue by class in each flight" : revenue_by_flight_class,
    "flights generating revenue > 50000 " : flight_generating_max_revenue,
    
    # bookings insights
    "passenger count in each flight" : passenger_count_per_flight,
    "most booked flight" : most_booked_flight,
    "cancellation rate" : cancellation_rate,
    
    #business intelligence
    "most profitable class" : most_profitable_class,
    "avg ticket price per flight" : flight_avg_price,
    
    #operational metrics
    "cancellation count per flight" : cancellation_cnt_per_flight,
    "total bookings per flight" : total_bookings_per_flight
    }

for i, (k,v) in enumerate(output_dict.items(), start=1):
    print(f"{i}. {k}\n : {v}\n")
```
