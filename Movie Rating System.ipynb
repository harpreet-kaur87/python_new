#### 🎬 Movie Rating System
📄 Problem Statement

You are given a dataset containing user ratings for different movies. Each record consists of a user ID, movie name, and the rating given by the user.

Your task is to analyze this dataset and compute the following:

1. Average Rating per Movie  
Calculate the average rating for each movie.  
2. Highest Rated Movie(s)  
Identify the movie(s) with the highest average rating.  
3. Active Users  
Find users who have rated more than 2 movies.  
4. Low Rated Movies  
List all movies whose average rating is below 3.

```python
ratings = [
    ("U1", "Movie1", 4),
    ("U2", "Movie1", 5),
    ("U1", "Movie2", 3),
    ("U3", "Movie2", 2),
    ("U2", "Movie3", 5),
    ("U1", "Movie3", 4),]


def movie_rating_system(lst):
    
    total = {}
    count = {}
    avg_rating_per_movie = {}
    user_count = {}
    
    
    for user,movie,rating in lst:
        
        if user not in user_count:
            user_count[user] = 0
        user_count[user] += 1
        
        if movie not in total:
            total[movie] = 0
            count[movie] = 0
        
        total[movie] += rating
        count[movie] += 1
        
    for movie in total:
        avg_rating_per_movie[movie] = total[movie]/count[movie]
        
    max_rating = max(avg_rating_per_movie.values())
    highest_rated_movie = [k for k,v in avg_rating_per_movie.items() if v == max_rating]
    
    users_rated_more = [k for k,v in user_count.items() if v > 2]
    
    below_avg_rating = [k for k,v in avg_rating_per_movie.items() if v < 3]
            
    return{ 
            "Avg rating per movie" : avg_rating_per_movie, 
            "Highest rated movie" : highest_rated_movie, 
        "Users who rated more than 2 movies" : users_rated_more, 
        "Movies with avg rating below 3" : below_avg_rating
        
    }
movie_rating_system(ratings)
```
